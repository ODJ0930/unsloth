### Save outputs and models to Google Drive

Links Studio's `outputs/` (checkpoints, LoRA weights) and `exports/` folders to `MyDrive/unsloth_studio/`, and puts the Hugging Face model cache at `MyDrive/unsloth_studio/huggingface/hub`, so results and downloaded models survive a Colab disconnect. Run it before starting Studio, in the same runtime session.

```python
# Persist training outputs and exports to Google Drive, so a Colab disconnect
# does not lose checkpoints. Re-run this cell after every reconnect, before starting Studio.
from google.colab import drive
from pathlib import Path
import os
import shutil

drive.mount("/content/drive")

DRIVE_ROOT = Path("/content/drive/MyDrive/unsloth_studio")  # change if you want another Drive folder
STUDIO_HOME = Path.home() / ".unsloth" / "studio"

for name in ("outputs", "exports"):
    target = DRIVE_ROOT / name
    target.mkdir(parents=True, exist_ok=True)
    local = STUDIO_HOME / name
    if local.is_symlink():
        local.unlink()
    elif local.exists():
        # Move anything already written on the VM into Drive before linking.
        for item in local.iterdir():
            dest = target / item.name
            if dest.exists():
                dest = target / f"{item.name}_from_vm"
            shutil.move(str(item), str(dest))
        local.rmdir()
    local.parent.mkdir(parents=True, exist_ok=True)
    local.symlink_to(target, target_is_directory=True)
    print(f"{local} -> {target}")

# Keep downloaded models on Drive too, so they are not re-downloaded after a reconnect.
# Studio honours an explicit HF_HUB_CACHE as long as it is set before Studio starts.
# HF_HOME (which holds your HF token) is left alone.
hf_cache = DRIVE_ROOT / "huggingface" / "hub"
hf_cache.mkdir(parents=True, exist_ok=True)
os.environ["HF_HUB_CACHE"] = str(hf_cache)
print(f"Model cache -> {hf_cache}")
```
